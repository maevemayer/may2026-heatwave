```python
import cdsapi, calendar, os

c = cdsapi.Client()
year, month = "2026", "05"
outdir = "."                                   # set your output folder
steps = [str(h) for h in range(0, 361, 6)]     # 6-hourly, 0–360 h
start_day = 10                                 # set to 1 for the whole month
area = [80, -130, 10, 80]                       # N, W, S, E
ndays = calendar.monthrange(int(year), int(month))[1]

os.makedirs(outdir, exist_ok=True)
failed = []

for d in range(start_day, ndays + 1):
    day = f"{d:02d}"
    fname = f"tigge_ecmwf_pf_2m_temperature_sfc_6h_{year}{month}{day}_05x05.grib"
    fpath = os.path.join(outdir, fname)

    if os.path.exists(fpath) and os.path.getsize(fpath) > 0:
        print(f"skip {fname} (already present)")
        continue

    tmp = fpath + ".tmp"
    try:
        c.retrieve("tigge-forecasts", {
            "origin": "ecmwf", "forecast_type": "perturbed_forecast",
            "year": year, "month": month, "day": day, "time": "00:00",
            "level_type": "surface", "variable": "2t",
            "leadtime_hour": steps,
            "area": area, "grid": "0.5/0.5", "data_format": "grib",
        }, tmp)
        os.replace(tmp, fpath)
        print(f"done {fname}")
    except Exception as e:
        if os.path.exists(tmp):
            os.remove(tmp)
        failed.append(day)
        print(f"FAILED {day}: {e}")

if failed:
    print("Re-run to retry these days:", failed)
else:
    print("All requested days present.")
```
